## Converting Categorical Columns into numerical:

```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
```

```python
cat_cols = ['niche', 'country']
encoded_data = encoder.fit_transform(df[cat_cols])
```

```python
encoded_df = pd.DataFrame(
    encoded_data, 
    columns=encoder.get_feature_names_out(cat_cols),
    index=df.index
)
```

```python
encoded_df
```

```python
df = pd.concat([df.drop(columns=cat_cols), encoded_df], axis=1)
```

```python
df
```

```python
df.isnull().sum()
# If we still doesnt get any missing values we're good to go.
```

```python
df.info()
```

```python
cols_to_numeric = [
    'stream_donations'
]
for col in cols_to_numeric:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
df.info()
```

## Correlation with our label:

```python
corr_mat = df.corr()
donation_corr = corr_mat['stream_donations'].sort_values(ascending=False)
```

```python
donation_corr
```

# Correlation Matrix Observations:

## 1. Positive Correlation (The Predictors)
These are the signals that suggest a stream is likely to earn more money. When these go up, stream_donations usually goes up.

### niche_Music (+0.16): 
This suggests individual music sessions are quite profitable.

### niche_IRL (+0.12): 
Real Life content is a safe for generating donations.

### chat_rate (+0.11): 
It proves that a "talkative" chat directly correlates with more money.

### like_count (+0.10): 
Streams that get more likes tend to see slightly higher donation totals.

### country_USA (+0.10): 
Streams coming from/targeting the USA are showing a consistent positive trend.

### niche_Finance (+0.07): 
Still in the positive zone.

## 2. Negative Correlation (The Inverse Signals)
Higher values here currently associate with lower donation totals in your specific dataset.

### avg_viewers (-0.22) & peak_viewers (-0.17): 
Massive audiences don't always mean massive donations; smaller, dedicated crowds often donate more per person.

### country_Canada (-0.22): 
At the stream level, it's showing a negative correlation.

### niche_Gaming (-0.18): 
Gaming continues to be a difficult niche for direct donations compared to others.

### niche_Tech (-0.11): 
Shows a slight negative trend at the individual stream level.

### duration_minutes (-0.10): 
This suggests that longer streams don't necessarily lead to more money; quality of time beats quantity of time.

## 3. Uncorrelated (The Noise)
These are very close to zero. The model will likely find these metrics "confusing" because they don't have a clear relationship with the money.

### country_Japan (+0.03): 
Almost no impact on the donation outcome.

### country_Germany (-0.01): 
Perfectly neutral; being in Germany doesn't seem to help or hurt donations here.

### niche_Sports (-0.04): 
Very low correlation, meaning it's not a strong predictor either way right now.
