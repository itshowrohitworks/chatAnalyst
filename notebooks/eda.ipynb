## Converting Categorical Columns into numerical:

```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
```

```python
cat_cols = ['niche', 'country']
encoded_data = encoder.fit_transform(df[cat_cols])
```

```python
encoded_df = pd.DataFrame(
    encoded_data, 
    columns=encoder.get_feature_names_out(cat_cols),
    index=df.index
)
```

```python
encoded_df
```

```python
df = pd.concat([df.drop(columns=cat_cols), encoded_df], axis=1)
```

```python
df
```

```python
df.isnull().sum()
# If we still doesnt get any missing values we're good to go.
```

```python
cols_to_numeric = [
    'avg_viewers', 'avg_peak_viewers', 'avg_chat_rate', 
    'avg_like_count', 'total_donations'
]
for col in cols_to_numeric:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
df.info()
```

## Correlation with our label:

```python
corr_mat = df.corr()
donation_corr = corr_mat['total_donations'].sort_values(ascending=False)
```

```python
donation_corr
```

# Correlation Matrix Observations:

## 1. Positive Correlations:
These features move in the same direction as your donations. When these numbers go up, the money usually goes up too. These are your strongest predictors.

### niche_IRL (+0.56): 
Niche has a very strong positive connection, suggesting that IRL content is currently the most profitable niche in your dataset.

### country_Canada (+0.31): 
Streams from Canada are strongly associated with higher donation totals.

### niche_Finance (+0.22): 
Finance-related content attracts viewers who are more likely to contribute financially.

### country_USA (+0.18): 
It is slightly less influential than Canada in this specific data batch.

### niche_Tech (+0.16): 
Tech content shows a healthy positive relationship with revenue.

### avg_chat_rate: 0.14
Streamer's engagement with it's chat drives his revenue.

## 2. Negative Correlation
These features move in the opposite direction. Surprisingly, in your data, higher numbers in these columns often mean lower total donations.

### niche_Sports (-0.40) & niche_Music (-0.39): 
For some reason, these niches are currently seeing the lowest donation totals per stream.

### country_Japan (-0.38) & country_India (-0.28): 
These geographic regions currently show an inverse relationship with the total donation amounts recorded.

### avg_peak_viewers (-0.28) & avg_viewers (-0.24): 
It suggests that as streams get massive, the donation total actually drops. Smaller, tighter communities are out-earning the huge crowds.

## 3. Uncorrelated
These features have scores very close to 0. They are neutral and don't seem to have any meaningful mathematical impact on donations right now.

### total_minutes (+0.04): 
Whether a stream is long or short doesn't seem to change how much money it makes.

### niche_Gaming (-0.01): 
Surprisingly, being a Gamer doesn't positively or negatively impact donations here; it's almost perfectly neutral.

### followers (-0.01): 
Having millions of followers doesn't guarantee more donations than a small streamer.

### avg_like_count (-0.06): 
Likes do look good but don't reliably predict how much money will be donated.
